```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score                          #Import all packages requested
```

```python
dataset=load_iris()             #load dataset into this code
```

```python
x_data=dataset.data
y_data=dataset.target           #assign question and answer data
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=24)     #split dataset into train(0,8) & test data(0.2), using random state #24
```

```python
neigh = KNeighborsClassifier(n_neighbors=3)     #declare using KNN algorithm and set k parameter to 3
neigh.fit(x_train,y_train)                      #run ML & train model
```

```python
y_pred = neigh.predict(x_test)                  #predict labels for test data
y_predict=accuracy_score(y_test,y_pred)         #evaluate model accuracy with accurate score
print(y_predict)                                #print model accuracy
```
